```python
import numpy as np
```

```python
def MSE(pred, true):
    return np.mean(np.square(true - pred))
```

```python
true = np.load("./pred_96.npy")
pred = np.random.rand(*true.shape)
mse1 = MSE(pred, true)

true = np.load("./pred_192.npy")
pred = np.random.rand(*true.shape)
mse2 = MSE(pred, true)

true = np.load("./pred_336.npy")
pred = np.random.rand(*true.shape)
mse3 = MSE(pred, true)

true = np.load("./pred_720.npy")
pred = np.random.rand(*true.shape)
mse4 = MSE(pred, true)

score = (mse1 + mse2 + mse3 + mse4) / 4

print(f"MSE Avg: {score:.4f}")
print(f"MSE 96: {mse1:.4f} || MSE 192: {mse2:.4f} || MSE 336: {mse3:.4f} || MSE 720: {mse4:.4f}")
```
